```python
# =================================================================================
# 🚀 EJECUCIÓN POST-AUTENTICACIÓN (Corre esto DESPUÉS de autenticarte)
# =================================================================================

# Inicializa el motor de Earth Engine
ee.Initialize()
print("\n--- 2. CONEXIÓN GEE EXITOSA ---")

# PRUEBA SIMPLE: Cargar una colección de imágenes pública (Landsat 8)
# y obtener el número total de imágenes.
try:
    # Colección Landsat 8 Nivel 2, de un período específico.
    landsat_collection = ee.ImageCollection('LANDSAT/LC08/C02/T1_L2')\
        .filterDate('2024-01-01', '2024-01-31')\
        .filterBounds(ee.Geometry.Point(-60.0, -34.0)) # Un punto en Argentina

    # Obtener el tamaño de la colección (la parte clave es .getInfo(), que ejecuta la consulta)
    count = landsat_collection.size().getInfo()

    print(f"✅ Consulta de prueba exitosa.")
    print(f"Imágenes Landsat 8 (Enero 2024) sobre la región de prueba: {count}")

except Exception as e:
    print(f"❌ Error al inicializar o ejecutar la consulta: {e}")


# PRUEBA DE VISUALIZACIÓN (Utilizando Folium, que instalaste)
# Creamos un mapa base simple de la zona de Buenos Aires.
mapa = folium.Map(location=[-34.6, -58.4], zoom_start=9)
print("\n--- 3. MAPA DE FOLIUM GENERADO ---")
mapa
```
